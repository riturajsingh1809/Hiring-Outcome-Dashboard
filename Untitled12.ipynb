```python
# ============================================================
# PROJECT : HIRING OUTCOME DASHBOARD
# PART - 1 : DATA LOADING + DATA CLEANING + DATA QUALITY CHECK
# ============================================================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Upload Dataset
from google.colab import files
uploaded = files.upload()

# Read Excel File
df = pd.read_excel("Sample Dataset.xlsx")

print("="*60)
print("DATASET LOADED SUCCESSFULLY")
print("="*60)

# First 5 Records
print("\nFirst 5 Records")
print(df.head())

# Last 5 Records
print("\nLast 5 Records")
print(df.tail())

# Shape
print("\nDataset Shape")
print(df.shape)

# Column Names
print("\nColumn Names")
print(df.columns)

# Data Types
print("\nData Types")
print(df.dtypes)

# Dataset Information
print("\nDataset Information")
df.info()

# Statistical Summary
print("\nStatistical Summary")
print(df.describe(include='all'))

# Missing Values
print("\nMissing Values")
print(df.isnull().sum())

# Total Missing Values
print("\nTotal Missing Values")
print(df.isnull().sum().sum())

# Duplicate Records
print("\nDuplicate Records")
print(df.duplicated().sum())

# Remove Duplicate Records
df = df.drop_duplicates()

print("\nShape After Removing Duplicates")
print(df.shape)

# Check Unique Values
print("\nUnique Values in Each Column")
print(df.nunique())

# Check Salary Statistics
if "Salary" in df.columns:
    print("\nSalary Statistics")
    print(df["Salary"].describe())

# Check Interview Status Count
if "Interview_Status" in df.columns:
    print("\nInterview Status")
    print(df["Interview_Status"].value_counts())

# Check Offer Status Count
if "Offer_Status" in df.columns:
    print("\nOffer Status")
    print(df["Offer_Status"].value_counts())

# Check Joining Status Count
if "Joining_Status" in df.columns:
    print("\nJoining Status")
    print(df["Joining_Status"].value_counts())

# Check Source Count
if "Source" in df.columns:
    print("\nHiring Source")
    print(df["Source"].value_counts())

# Convert Date Columns
if "Applied_Date" in df.columns:
    df["Applied_Date"] = pd.to_datetime(df["Applied_Date"])

if "Interview_Date" in df.columns:
    df["Interview_Date"] = pd.to_datetime(df["Interview_Date"])

print("\nDate Columns Converted Successfully")

# Save Cleaned Dataset
df.to_csv("cleaned_hiring_dataset.csv", index=False)

print("\nCleaned Dataset Saved Successfully")

print("="*60)
print("PART - 1 COMPLETED SUCCESSFULLY")
print("="*60)
```

```python
# ============================================================
# PROJECT : HIRING OUTCOME DASHBOARD
# PART - 2 : KPI ANALYSIS & HIRING INSIGHTS
# ============================================================

print("="*60)
print("HIRING OUTCOME DASHBOARD - KPI ANALYSIS")
print("="*60)

# -----------------------------
# Basic KPIs
# -----------------------------

total_candidates = len(df)

selected_candidates = (df["Interview_Status"] == "Selected").sum()

rejected_candidates = (df["Interview_Status"] == "Rejected").sum()

offers_sent = (df["Offer_Status"] == "Sent").sum()

offers_signed = (df["Offer_Signed"] == "Yes").sum()

joined_candidates = (df["Joining_Status"] == "Joined").sum()

pending_joining = (df["Joining_Status"] == "Pending").sum()

average_salary = df["Salary"].mean()

maximum_salary = df["Salary"].max()

minimum_salary = df["Salary"].min()

# -----------------------------
# Print KPIs
# -----------------------------

print(f"Total Candidates        : {total_candidates}")
print(f"Selected Candidates     : {selected_candidates}")
print(f"Rejected Candidates     : {rejected_candidates}")
print(f"Offers Sent             : {offers_sent}")
print(f"Offers Signed           : {offers_signed}")
print(f"Joined Candidates       : {joined_candidates}")
print(f"Pending Joining         : {pending_joining}")
print(f"Average Salary          : {average_salary:.2f}")
print(f"Maximum Salary          : {maximum_salary}")
print(f"Minimum Salary          : {minimum_salary}")

# -----------------------------
# Position Wise Hiring
# -----------------------------

print("\nPosition Wise Candidates")
print(df["Position"].value_counts())

# -----------------------------
# Hiring Source Analysis
# -----------------------------

print("\nHiring Source Analysis")
print(df["Source"].value_counts())

# -----------------------------
# Interview Status
# -----------------------------

print("\nInterview Status")
print(df["Interview_Status"].value_counts())

# -----------------------------
# Offer Signed Analysis
# -----------------------------

print("\nOffer Signed Analysis")
print(df["Offer_Signed"].value_counts())

# -----------------------------
# Joining Status Analysis
# -----------------------------

print("\nJoining Status")
print(df["Joining_Status"].value_counts())

# -----------------------------
# Average Salary by Position
# -----------------------------

print("\nAverage Salary by Position")
print(df.groupby("Position")["Salary"].mean().round(2))

# -----------------------------
# Average Salary by Source
# -----------------------------

print("\nAverage Salary by Hiring Source")
print(df.groupby("Source")["Salary"].mean().round(2))

# -----------------------------
# Final KPI Summary Table
# -----------------------------

summary = pd.DataFrame({
    "KPI": [
        "Total Candidates",
        "Selected Candidates",
        "Rejected Candidates",
        "Offers Sent",
        "Offers Signed",
        "Joined Candidates",
        "Pending Joining",
        "Average Salary"
    ],
    "Value": [
        total_candidates,
        selected_candidates,
        rejected_candidates,
        offers_sent,
        offers_signed,
        joined_candidates,
        pending_joining,
        round(average_salary,2)
    ]
})

print("\nKPI SUMMARY")
print(summary)

# Save KPI Summary
summary.to_csv("kpi_summary.csv", index=False)

print("\nKPI Summary Saved Successfully!")

print("="*60)
print("PART - 2 COMPLETED SUCCESSFULLY")
print("="*60)
```

```python
# ============================================================
# PROJECT : HIRING OUTCOME DASHBOARD
# PART - 3 : CHARTS & EXECUTIVE DASHBOARD
# ============================================================

import matplotlib.pyplot as plt

print("="*60)
print("CREATING HIRING DASHBOARD")
print("="*60)

# ============================================================
# Chart 1 : Interview Status
# ============================================================

plt.figure(figsize=(6,4))
df["Interview_Status"].value_counts().plot(kind="bar", color=["green","red"])
plt.title("Interview Status")
plt.xlabel("Status")
plt.ylabel("Number of Candidates")
plt.grid(axis='y')
plt.savefig("interview_status.png")
plt.show()

# ============================================================
# Chart 2 : Hiring Source
# ============================================================

plt.figure(figsize=(6,6))
df["Source"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.ylabel("")
plt.title("Hiring Source Distribution")
plt.savefig("hiring_source.png")
plt.show()

# ============================================================
# Chart 3 : Position Wise Hiring
# ============================================================

plt.figure(figsize=(7,4))
df["Position"].value_counts().plot(kind="bar")
plt.title("Position Wise Hiring")
plt.xlabel("Position")
plt.ylabel("Candidates")
plt.grid(axis='y')
plt.savefig("position_wise_hiring.png")
plt.show()

# ============================================================
# Chart 4 : Joining Status
# ============================================================

plt.figure(figsize=(6,4))
df["Joining_Status"].value_counts().plot(kind="bar")
plt.title("Joining Status")
plt.xlabel("Status")
plt.ylabel("Candidates")
plt.grid(axis='y')
plt.savefig("joining_status.png")
plt.show()

# ============================================================
# Chart 5 : Salary Distribution
# ============================================================

plt.figure(figsize=(7,4))
plt.hist(df["Salary"], bins=8)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.grid()
plt.savefig("salary_distribution.png")
plt.show()

# ============================================================
# Chart 6 : Offer Signed
# ============================================================

plt.figure(figsize=(6,6))
df["Offer_Signed"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.ylabel("")
plt.title("Offer Signed")
plt.savefig("offer_signed.png")
plt.show()

# ============================================================
# Dashboard KPIs
# ============================================================

print("\n========== EXECUTIVE DASHBOARD ==========\n")

print("Total Candidates :", len(df))
print("Selected Candidates :", (df["Interview_Status"]=="Selected").sum())
print("Rejected Candidates :", (df["Interview_Status"]=="Rejected").sum())
print("Offers Sent :", (df["Offer_Status"]=="Sent").sum())
print("Offers Signed :", (df["Offer_Signed"]=="Yes").sum())
print("Joined Candidates :", (df["Joining_Status"]=="Joined").sum())
print("Average Salary :", round(df["Salary"].mean(),2))

print("\nTop Hiring Source")
print(df["Source"].value_counts())

print("\nTop Hiring Position")
print(df["Position"].value_counts())

print("\nDashboard Images Saved Successfully")

print("""
Saved Files
-----------
interview_status.png
hiring_source.png
position_wise_hiring.png
joining_status.png
salary_distribution.png
offer_signed.png
""")

print("="*60)
print("PART - 3 COMPLETED SUCCESSFULLY")
print("="*60)
```
